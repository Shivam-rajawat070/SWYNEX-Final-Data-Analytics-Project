Import Libraries and Load Data

```python
import pandas as pd
import numpy as np

data = pd.read_csv("C:/Users/shiva/Downloads/raw_sales_dataset_2000.csv")
print(data)
```

 Explore the Dataset Structure

```python
print(data.shape)
data.info()
```

 Check for Missing Values

```python
data.isnull().sum()
```

 Check for Duplicate Records

```python
data.duplicated().sum()
```

 Check for Inconsistent Values

```python
print(data['City'].unique())
print(data['Rating'].unique())
```

 Remove Duplicate Records

```python
data = data.drop_duplicates()
print(data.shape)
```

 Fix Inconsistent City Names

```python
data['City'] = data['City'].str.strip().str.lower()

data['City'] = data['City'].replace({
    'bengaluru': 'bangalore',
    'new delhi': 'delhi'
})

data['City'] = data['City'].str.title()
print(data['City'].unique())
```

 Fix Inconsistent Rating Values

```python
data['Rating'] = data['Rating'].replace('five', '5')
data['Rating'] = pd.to_numeric(data['Rating'], errors='coerce')
print(data['Rating'].unique())
```

 Fix OrderDate Data Type

```python
data['OrderDate'] = pd.to_datetime(data['OrderDate'], format='mixed', dayfirst=True, errors='coerce')
print(data['OrderDate'].head(10))
print(data['OrderDate'].isnull().sum())
```

 Handle Missing Values in Numeric Columns

```python
data['Quantity'] = data['Quantity'].fillna(data['Quantity'].median())
data['Price'] = data['Price'].fillna(data['Price'].median())
data['Rating'] = data['Rating'].fillna(data['Rating'].median())
print(data.isnull().sum())
```

 Handle Missing Values in Text Columns

```python
data['PaymentMode'] = data['PaymentMode'].fillna(data['PaymentMode'].mode()[0])
data['CustomerName'] = data['CustomerName'].fillna('Unknown')
print(data.isnull().sum())
```

 Re-check for Duplicates After Cleaning

```python
data = data.drop_duplicates()
print(data.duplicated().sum())
print(data.shape)
```

 Final Verification

```python
print(data.isnull().sum())
print(data.duplicated().sum())
data.info()
```

 Save the Cleaned Dataset

```python
data.to_csv('cleaned_sales_dataset.csv', index=False)
print("Saved successfully!")
```
